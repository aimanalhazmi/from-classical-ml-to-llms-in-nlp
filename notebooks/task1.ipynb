# Report


### **Task 1.1: Data Selection and Preprocessing**

#### **1. Language Selection & Rationale**

For this study, we selected three languages to investigate cross-lingual transfer capabilities in medical social media analysis:

* **English (`en`):** Selected as the primary high-resource source language ( raw samples).
* **Russian (`ru`):** Selected as a secondary source language (). Including Russian adds linguistic diversity (Cyrillic script, rich morphology), preventing the model from overfitting to Germanic/Latin patterns.
* **German (`de`):** Selected as the **Zero-Shot Target** language. It represents a low-resource scenario in this dataset ( samples) with a very low number of positive cases (only 83 positives). This makes it an ideal candidate for evaluating how well models trained on En/Ru generalize to unseen languages.

#### **2. Dataset Statistics & Class Imbalance**

A major challenge identified in the dataset is severe **class imbalance**. As shown in Table 1, the ratio of positive samples (Adverse Drug Events) is consistently low across all languages.

| Language | Total Samples (Train/Val) | Positive (Class 1) | Negative (Class 0) | Positive Ratio |
| --- | --- | --- | --- | --- |
| **English (`en`)** | 17,128 | 1,198 | 15,930 | **7.0%** |
| **Russian (`ru`)** | 10,695 | 1,082 | 9,613 | **10.1%** |
| **German (`de`)** | 1,482 | 83 | 1,399 | **5.6%** |
| **TOTAL** | **29,305** | **2,363** | **26,942** | **8.1%** |

*Observation:* The model will likely bias towards the negative class. We addressed this during the train/validation split by performing **Stratified Splitting** (stratified by `Language + Label`). This ensured that the validation set maintained the exact same positive ratios (En: 7.0%, Ru: 10.1%) as the training set, preventing skewed evaluation results.

#### **3. Preprocessing Strategy**

We implemented a custom preprocessing pipeline tailored for SMM4H (Social Media Mining for Health). The impact of each step on the training data is quantified below:

* **User Anonymization (@User  `[USER]`):**
* *Impact:* Affected **7,942** records (~27% of data).
* *Justification:* This was the most frequent noise pattern. Anonymization is crucial to strictly adhere to privacy standards and to prevent the model from "overfitting" to specific active users (learning that User X is sick, rather than learning the symptoms).


* **Hashtag Cleaning (`#flu`  `flu`):**
* *Impact:* Affected **3,287** records.
* *Justification:* Hashtags often contain the most important semantic content (keywords). Removing the `#` symbol allows the tokenizer to treat them as standard words, preserving the medical context.


* **HTML & Artifact Removal:**
* *Impact:* Affected **2,281** records.
* *Justification:* Tags like `<br>` break sentence structure. We replaced them with periods (`.`) to restore sentence boundaries for the Transformer model.


* **Emoji Handling (🤢  `:nauseated_face:`):**
* *Impact:* Affected **478** records.
* *Justification:* Emojis are dense information carriers in social media (e.g., a pill or vomit face). Demojizing them into text ensures the BERT model interprets the *sentiment* rather than treating them as unknown tokens.


* **Capitalization (Case Preservation):**
* *Decision:* We explicitly **kept** original capitalization.
* *Justification:* Medical acronyms (e.g., `AIDS`, `ALL`, `US`) are case-sensitive. Lowercasing them would introduce ambiguity (e.g., "aids" as in hearing aids vs. the disease).



#### **4. Final Split Construction**

After preprocessing, the data was split 80/20 into `train.csv` and `val.csv`.

* **Final Train Set:** 23,395 samples (Contains En, Ru, De).
* **Final Val Set:** 5,849 samples.
* **Note:** For the main Task 1 training, we will filter the `train.csv` to strictly exclude German (`de`), reserving it solely for Zero-Shot evaluation.

### **Task 1.2: Exploratory Data Analysis & Linguistic Characterization**

#### **1. Dataset Distribution & Balance**

We analyzed the class distribution across the three target languages (`en`, `ru`, `de`) in the training, validation, and test sets.

* **Class Imbalance:**
The dataset exhibits a severe class imbalance. As shown in the `Label Distribution` table, the positive class (Adverse Drug Events) accounts for only **8.1%** of the training data (1,891 positives vs. 21,504 negatives).
* **Russian (`ru`)** has the highest density of positives (**10.1%**).
* **German (`de`)** has the lowest density (**5.6%**), making it a challenging Zero-Shot target due to the scarcity of positive examples even in the source distribution.


* **Language Imbalance:**
English dominates the training corpus (**58.4%**), followed by Russian (**36.6%**). German is a low-resource minority (**5.1%**). This imbalance suggests that a multilingual model will likely bias its internal representations toward English syntax and vocabulary unless addressed (e.g., via resampling or balanced batching).

#### **2. Source Type Analysis**

The data originates from diverse social media platforms, which impacts the linguistic style:

* **Tweets (80.8%):** Short, informal, often containing hashtags and user mentions.
* **Sentences (12.9%):** Likely extracted from longer documents; these tend to be more grammatically standard.
* **Forum Posts (5.1%):** Longer, narrative-style descriptions of personal experiences.

#### **3. Linguistic Characteristics**

We analyzed token lengths (whitespace-split) to understand input complexity.

* **Sequence Lengths:**
* **English & Russian:** Similar average lengths (~16-19 tokens).
* **German:** Significantly longer (**Avg: 100.1 tokens**, Max: 984).
* *Interpretation:* The German subset likely consists of **Forum Posts** rather than Tweets. This drastic difference in length means the model must learn to handle both short, noisy tweets (En/Ru) and long, narrative paragraphs (De). Zero-shot transfer might struggle not just with language, but with the **shift in discourse structure** (short vs. long context).


* **Length vs. Label:**
Positive samples (ADE) are on average **longer** (31.9 tokens) than negative samples (20.9 tokens). This suggests that users describing a side effect tend to provide more narrative detail ("I took X and then Y happened") compared to simple statements ("I need Xanax").

#### **4. Manual Inspection & Qualitative Analysis**

We randomly sampled entries to identify potential challenges:

* **Ambiguity in Negatives:**
* *Example (En):* "I need some ambien or geodon to knock my ass out." -> Labeled **0**.
* *Analysis:* Mentioning a drug for its *intended* purpose (sedation) is not an ADE, but a model might confuse the sentiment "knock my ass out" with a negative effect.


* **Cultural/Linguistic Nuances:**
* *Example (Ru):* A user describes "gnawing nails" due to anxiety and how "Afobazol" helped. This narrative style is complex and requires the model to understand *relief* of symptoms vs. *causation* of symptoms.


* **Translation Quality (En  De):**
* *Source:* "doing Prolia arms."
* *Translation:* "...bekomme Prolia in die Arme."
* *Artifact:* The translation is literal. "Doing Prolia arms" likely refers to an injection site or a specific sensation, which the literal German translation might miss contextually. However, drug names (`Neurontin`, `Paxil`) were correctly preserved.



#### **5. Conclusion**

The EDA reveals three critical challenges for the Zero-Shot task:

1. **Extreme Imbalance:** The target language (De) has very few positives (5.6%).
2. **Domain Shift:** German data is much longer (forum posts) than the source training data (mostly tweets).
3. **Linguistic Diversity:** The model must bridge the gap between English/Russian morphology and German compound structures while handling the shift from short-text to long-text classification.

```python
import pandas as pd
from datasets import DatasetDict, Dataset
from src.utils import get_project_root
def load_data(cfg: dict):
    """
    Loads data from CSVs specified in config.
    Filters by language if 'lang_filter' is provided.
    """
    ds_cfg = cfg["dataset"]
    project_root = get_project_root()

    datasets = {}
    for split in ["train", "val", "test"]:
        path = project_root / ds_cfg[f"{split}_path"]
        if path.exists():
            df = pd.read_csv(path)

            # Ensure text is string and handle NaNs
            df[ds_cfg["text_column"]] = df[ds_cfg["text_column"]].fillna("").astype(str)

            # Filter Languages (Train only on en and ru)
            if "lang_filter" in ds_cfg and split == "train":
                selected_langs = ds_cfg["lang_filter"]
                print(f"Filtering {split} set to languages: {selected_langs}")
                df = df[df['language'].isin(selected_langs)]

            # Rename columns to standard HF names
            df = df.rename(columns={ds_cfg["text_column"]: "text", ds_cfg["label_column"]: "labels"})

            datasets[split] = Dataset.from_pandas(df)
            print(f"Loaded {split}: {len(df)} rows")
        else:
            print(f"[red]Warning: {split} file not found at {path}[/red]")

    return DatasetDict(datasets)
```

```python
cfg = {"dataset": {
    "train_path": "data/SMM4H_2026/train.csv",
    "val_path": "data/SMM4H_2026/val.csv",
    "test_path": "data/SMM4H_2026/test.csv",
    "text_column": "text",
    "label_column": "label",
    "lang_filter": ["en", "ru"]
  }}
data = load_data(cfg=cfg)
```
